## Segmentação de Dataframes

### Loc e Iloc

```python
base
```

```python
# Não segue a regra geral : Primeiro número incluso e último número também está incluso
# Porque estamos chamando rótulos, e não números

base.loc[0:5,['PassengerId','Pclass']]
```

```python
# Segue a regra geral: Primeiro número incluso, e último número não incluso
# Porque estamos chamando índices/números e não rótulos

base.iloc[-4:,-3:]
```

```python
base[0:5]
```

```python
base['PassengerId']
```
